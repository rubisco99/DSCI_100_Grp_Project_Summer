#Exploratory Data Analysis (EDA)

##1. Data Cleaning & Sample Selection

```r
df_eda <- dbGetQuery(con, "
  SELECT
    g.game_id,
    g.game_date,
    g.season_id,

    -- Home Team Scoring (Response Variables)
    g.team_id_home AS home_team_id,
    g.pts_home AS home_pts,
    CASE WHEN g.pts_home > 100 THEN 1 ELSE 0 END AS high_scoring_100,

    -- Away Team Context (Covariates)
    g.team_id_away AS away_team_id,
    g.pts_away AS away_pts
  FROM game g
  WHERE g.pts_home IS NOT NULL
  LIMIT 5000
")
#5000 games only

head(df_eda)
```

## 2. Summary Statistics and Univariate Analysis

```r
summary_stats <- df_eda %>%
  reframe(
    Variable = c("Home Points (home_pts)", "Away Points (away_pts)"),
    Mean     = c(mean(home_pts), mean(away_pts)),
    SD       = c(sd(home_pts), sd(away_pts)),
    Median   = c(median(home_pts), median(away_pts)),
    IQR      = c(IQR(home_pts), IQR(away_pts)),
    Min      = c(min(home_pts), min(away_pts)),
    Max      = c(max(home_pts), max(away_pts))
  )

  summary_stats

  # Note: Away Points here means opponent points
```
